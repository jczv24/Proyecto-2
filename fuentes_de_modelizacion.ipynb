```python
def train_model(parameters, train_X, train_y, test_X, test_y, experiment_name, early_stop):
    
    model = Sequential()
    model.add(Input(shape=(train_X.shape[1],)))
    model.add(Dense(parameters['hidden_units'], activation = parameters['act_func']))
    
    if parameters['layers'] > 1:
        model.add(Dense(parameters['hidden_units'], activation = parameters['act_func']))
    if parameters['layers'] == 3:
        model.add(Dense(parameters['hidden_units'], activation = parameters['act_func']))
        
    model.add(Dense(1, activation='sigmoid'))
    
    if parameters['optimizer'] == 'adam':
        opt = keras.optimizers.Adam(learning_rate = parameters['learning_rate'])
    elif parameters['optimizer'] == 'adagrad':
        opt = keras.optimizers.Adagrad(learning_rate = parameters['learning_rate'])
    elif parameters['optimizer'] == 'sgd':
        opt = keras.optimizers.SGD(learning_rate = parameters['learning_rate'])
    
    model.compile(optimizer = opt, loss = 'binary_crossentropy', metrics = ['accuracy', 'true_positives', 'true_negatives', 'false_positives', 'false_negatives'])
    
    # defina el servidor para llevar el registro de modelos y artefactos
    mlflow.set_tracking_uri('http://127.0.0.1:5000')
    # registre el experimento
    experiment = mlflow.set_experiment(experiment_name)
    
    # Aquí se ejecuta MLflow sin especificar un nombre o id del experimento. MLflow los crea un experimento para este cuaderno por defecto y guarda las características del experimento y las métricas definidas. 
    # Para ver el resultado de las corridas haga click en Experimentos en el menú izquierdo. 
    with mlflow.start_run(experiment_id=experiment.experiment_id):
        
        history = model.fit(train_X, train_y,
                  epochs = parameters['epochs'],
                  batch_size = parameters['batch_size'],
                  validation_split = 0.2,
                  callbacks = [early_stop],
                  verbose = 0
                  )
        
        #Evaluar el modelo
        eval_model = model.evaluate(test_X, test_y, verbose = 0)
        loss = eval_model[0]
        accuracy = eval_model[1]
        tp = eval_model[2]
        tn = eval_model[3]
        fp = eval_model[4]
        fn = eval_model[5]
        
        pos_precision = tp / (tp + fp)
        neg_precision = tn / (tn + fn)
        pos_recall = tp / (tp + fn)
        neg_recall = tn / (tn + fp)
        error_I = fp / (fp + tn)
        error_II = fn / (fn + tp)
        
        # Log parameters and results
        mlflow.log_params(parameters)
        mlflow.log_metric("accuracy", round(accuracy, 5))
        mlflow.log_metric("test_loss", round(loss, 5))
        mlflow.log_metric("positive_precision", round(pos_precision, 5))
        mlflow.log_metric("negative_precision", round(neg_precision, 5))
        mlflow.log_metric("positive_recall", round(pos_recall, 5))
        mlflow.log_metric("negative_recall", round(neg_recall, 5))
        mlflow.log_metric("Error_I", round(error_I, 5))
        mlflow.log_metric("Error_II", round(error_II, 5))
        print(f'Loss: {loss}, Accuracy: {accuracy}')
        
        
        
        # cm = confusion_matrix(y_test, y_pred)
        # pre_bajo = cm[0, 0] / (cm[0, 0] + cm[1, 0] + cm[2, 0])
        # pre_medio = cm[1, 1] / (cm[0, 1] + cm[1, 1] + cm[2, 1])
        # pre_alto = cm[2, 2] / (cm[0, 2] + cm[1, 2] + cm[2, 2])
        # rec_bajo = cm[0, 0] / (cm[0, 0] + cm[0, 1] + cm[0, 2])
        # rec_medio = cm[1, 1] / (cm[1, 0] + cm[1, 1] + cm[1, 2])
        # rec_alto = cm[2, 2] / (cm[2, 0] + cm[2, 1] + cm[2, 2])
        # tot_bajo = cm[0, 0] + cm[0, 1] + cm[0, 2]
        # tot_medio = cm[1, 0] + cm[1, 1] + cm[1, 2]
        # tot_alto = cm[2, 0] + cm[2, 1] + cm[2, 2]
        # tot = tot_bajo + tot_medio + tot_alto
        # wa_pre = ((tot_bajo * pre_bajo) + (tot_medio * pre_medio) + (tot_alto * pre_alto)) / tot
        # wa_rec = ((tot_bajo * rec_bajo) + (tot_medio * rec_medio) + (tot_alto * rec_alto)) / tot
        
        
        # mlflow.log_metric("precision_bajo", round(pre_bajo, 5))
        # mlflow.log_metric("precision_medio", round(pre_medio, 5))
        # mlflow.log_metric("precision_alto", round(pre_alto, 5))
        # mlflow.log_metric("recall_bajo", round(rec_bajo, 5))
        # mlflow.log_metric("recall_medio", round(rec_medio, 5))
        # mlflow.log_metric("recall_alto", round(rec_alto, 5))
        # mlflow.log_metric("weighted_precision", round(wa_pre, 5))
        # mlflow.log_metric("weighted_recall", round(wa_rec, 5))
        
        # Log model
        mlflow.tensorflow.log_model(model, "model")
            
    
        return {"model": model, 'history' : history}
```

```python
def show_history(history):
    plt.plot(hist.history['loss'], label='Entrenamiento')
    plt.plot(hist.history['val_loss'], label='Validación')
    plt.xlabel('Época')
    plt.ylabel('Precisión')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
def param_tuning(param_grid, nombre_prueba):
    earlystopping = callbacks.EarlyStopping(monitor = "val_loss",
                                        mode = "min",
                                        patience = 5,
                                        restore_best_weights = True)
    
    mod_hist = []
    for params in param_grid:
        models = []
        modelo = train_model(params, X_train, y_train, X_test, y_test, nombre_prueba, earlystopping)
        models.append(modelo['model'])
        models.append(modelo['history'])
        mod_hist.append(models)
    
    return mod_hist
```

### Prueba de modelos con dos capas y cambiando el número de neuronas y funciones de activación por capa

```python
def train_model(parameters, train_X, train_y, test_X, test_y, experiment_name, early_stop):
    
    model = Sequential()
    model.add(Input(shape=(train_X.shape[1],)))
    model.add(Dense(parameters['hidden_units_1'], activation = parameters['act_func_1']))
    model.add(Dense(parameters['hidden_units_2'], activation = parameters['act_func_2']))
        
    model.add(Dense(3, activation='softmax'))
    
    if parameters['optimizer'] == 'adam':
        opt = keras.optimizers.Adam(learning_rate = parameters['learning_rate'])
    elif parameters['optimizer'] == 'adagrad':
        opt = keras.optimizers.Adagrad(learning_rate = parameters['learning_rate'])
    elif parameters['optimizer'] == 'sgd':
        opt = keras.optimizers.SGD(learning_rate = parameters['learning_rate'])
    
    model.compile(optimizer = opt, loss = 'categorical_crossentropy', metrics = ['accuracy'])
    
    # defina el servidor para llevar el registro de modelos y artefactos
    mlflow.set_tracking_uri('http://127.0.0.1:5000')
    # registre el experimento
    experiment = mlflow.set_experiment(experiment_name)
    
    # Aquí se ejecuta MLflow sin especificar un nombre o id del experimento. MLflow los crea un experimento para este cuaderno por defecto y guarda las características del experimento y las métricas definidas. 
    # Para ver el resultado de las corridas haga click en Experimentos en el menú izquierdo. 
    with mlflow.start_run(experiment_id=experiment.experiment_id):
        
        history = model.fit(train_X, train_y,
                  epochs = parameters['epochs'],
                  batch_size = parameters['batch_size'],
                  validation_split = 0.2,
                  callbacks = [early_stop],
                  verbose = 0
                  )
        
        #Evaluar el modelo
        eval_model = model.evaluate(test_X, test_y, verbose = 0)
        loss = eval_model[0]
        accuracy = eval_model[1]
        # Log parameters and results
        mlflow.log_params(parameters)
        mlflow.log_metric("accuracy", round(accuracy, 5))
        mlflow.log_metric("test_loss", round(loss, 5))
        print(f'Loss: {loss}, Accuracy: {accuracy}')
        
        #Valores de predicción (Precision & Recall)
        y_pred = model.predict(test_X)
        y_pred = np.argmax(y_pred, axis=1)
        y_test = np.argmax(test_y, axis=1)
        
        cm = confusion_matrix(y_test, y_pred)
        pre_bajo = cm[0, 0] / (cm[0, 0] + cm[1, 0] + cm[2, 0])
        pre_medio = cm[1, 1] / (cm[0, 1] + cm[1, 1] + cm[2, 1])
        pre_alto = cm[2, 2] / (cm[0, 2] + cm[1, 2] + cm[2, 2])
        rec_bajo = cm[0, 0] / (cm[0, 0] + cm[0, 1] + cm[0, 2])
        rec_medio = cm[1, 1] / (cm[1, 0] + cm[1, 1] + cm[1, 2])
        rec_alto = cm[2, 2] / (cm[2, 0] + cm[2, 1] + cm[2, 2])
        tot_bajo = cm[0, 0] + cm[0, 1] + cm[0, 2]
        tot_medio = cm[1, 0] + cm[1, 1] + cm[1, 2]
        tot_alto = cm[2, 0] + cm[2, 1] + cm[2, 2]
        tot = tot_bajo + tot_medio + tot_alto
        wa_pre = ((tot_bajo * pre_bajo) + (tot_medio * pre_medio) + (tot_alto * pre_alto)) / tot
        wa_rec = ((tot_bajo * rec_bajo) + (tot_medio * rec_medio) + (tot_alto * rec_alto)) / tot
        
        
        mlflow.log_metric("precision_bajo", round(pre_bajo, 5))
        mlflow.log_metric("precision_medio", round(pre_medio, 5))
        mlflow.log_metric("precision_alto", round(pre_alto, 5))
        mlflow.log_metric("recall_bajo", round(rec_bajo, 5))
        mlflow.log_metric("recall_medio", round(rec_medio, 5))
        mlflow.log_metric("recall_alto", round(rec_alto, 5))
        mlflow.log_metric("weighted_precision", round(wa_pre, 5))
        mlflow.log_metric("weighted_recall", round(wa_rec, 5))
        
        # Log model
        mlflow.tensorflow.log_model(model, "model")
            
    
        return {"model": model, 'history' : history}
```
